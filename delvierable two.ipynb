```python
print("Cholesterol Values:", dnkn['Cholesterol (mg)'].unique())
print("Trans Fat Values:", dnkn['Trans Fat (g)'].unique())
```

```python
dnkn1 = dnkn.copy()
```

```python
food_items = ['Donuts', 'Muffins', 'Kolache', 'Sandwiches', 'Hash Browns']
dnkn1 = dnkn1[~dnkn1['Category'].isin(food_items)].copy()
```

```python
dnkn1.duplicated().sum()
```

```python
dnkn2 = dnkn1.copy()
dnkn2['Cholesterol (mg)'] = dnkn2['Cholesterol (mg)'].replace('<5', '5')

for col in dnkn2.select_dtypes(include='object').columns:
    dnkn2[col] = (
        dnkn2[col]
        .str.replace('\xa0', '', regex=False)
        .str.strip()
    )
```

```python
for col in dnkn2.columns:
    if dnkn2[col].dtype == 'object':
        print('\n', col)
        print(dnkn2[col].unique()[:50])
```

```python
dnkn2 = dnkn2.replace('', np.nan)
```

```python
dnkn2.isnull().sum()
```

```python
dnkn3 = dnkn2.dropna().copy()
```

```python
dnkn3.isnull().sum()
```

```python
numeric_columns = [
    'Calories', 'Total Fat (g)', 'Saturated Fat (g)', 'Trans Fat (g)', 
    'Cholesterol (mg)', 'Sodium (mg)', 'Carbs (g)', 'Fiber (g)', 
    'Sugars (g)', 'Protein (g)', 'Weight Watcher Pnts'
]

for col in numeric_columns:
    dnkn3[col] = pd.to_numeric(dnkn3[col])
```

```python
print("================ MISSING VALUES ================")
print(dnkn3.isnull().sum())

print("\n================ DATA TYPES ================")
print(dnkn3.dtypes)

print("\n================ DUPLICATE RECORDS ================")
print("Duplicates:", dnkn3.duplicated().sum())

print("\n================ CATEGORY CHECK ================")
print(dnkn3['Category'].value_counts())

print("\n================ DATASET SHAPE ================")
print(dnkn3.shape)

print("\n================ SUMMARY STATISTICS ================")
print(dnkn3.describe().round(2))
```

```python
dnkn3.to_csv('dnkn_clean.csv', index=False)
```
